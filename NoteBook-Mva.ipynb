# ORM
Génération d'un MCD à partir des 3 fichiers voir MCD.drawio.
Cela permet de générer un script sql de création de la base :

Voici un script SQL généré à partir du Modèle Conceptuel de Données (MCD) fourni dans les sources
. Ce script définit les tables, les colonnes, les types de données et les contraintes d'intégrité référentielle (clés étrangères).
-- Création de la table Machine
-- Table Machine
CREATE TABLE Machine (
    ID BIGINT PRIMARY KEY NOT NULL, -- PK long not null
    Nom VARCHAR(255) NOT NULL       -- string not null
);

-- Table Operateur
CREATE TABLE Operateur (
    ID BIGINT PRIMARY KEY NOT NULL, -- PK long not null
    Nom VARCHAR(255) NOT NULL       -- string not null
);

-- Table Capteur (Liée à Machine)
CREATE TABLE Capteur (
    ID BIGINT PRIMARY KEY NOT NULL, -- PK long not null
    Nom VARCHAR(255) NOT NULL,      -- string not null
    Type INTEGER NULL,              -- int null
    Unit VARCHAR(50) NULL,          -- string null
    Machine_ID BIGINT NOT NULL,
    CONSTRAINT fk_machine FOREIGN KEY (Machine_ID) REFERENCES Machine(ID)
);

-- Table Incident (Liée à Machine et Operateur)
-- Changement : Lien direct avec Machine 1 <-> 0..n [1]
CREATE TABLE Incident (
    ID BIGINT PRIMARY KEY NOT NULL,   -- PK long not null
    Date TIMESTAMP NOT NULL,          -- datetime not null
    severite SMALLINT NULL,           -- short null
    Commentaire TEXT NULL,            -- string null
    quarter SMALLINT NULL,            -- short null
    Machine_ID BIGINT NOT NULL,
    Operateur_ID BIGINT NOT NULL,
    CONSTRAINT fk_machine_incident FOREIGN KEY (Machine_ID) REFERENCES Machine(ID),
    CONSTRAINT fk_operateur_incident FOREIGN KEY (Operateur_ID) REFERENCES Operateur(ID)
);

-- Table Mesure (Liée à Capteur)
CREATE TABLE Mesure (
    ID BIGINT PRIMARY KEY NOT NULL,   -- PK long not null
    Valeur DOUBLE PRECISION NULL,     -- double null
    Date TIMESTAMP NULL,              -- datetime null
    Capteur_ID BIGINT NOT NULL,
    CONSTRAINT fk_capteur_mesure FOREIGN KEY (Capteur_ID) REFERENCES Capteur(ID)
);
Détails de l'implémentation basés sur le MCD
 :
Clés Primaires (PK) : Toutes les entités possèdent un identifiant de type long (traduit ici par BIGINT)
.
Gestion des relations :
La table Capteur inclut une clé étrangère vers Machine car une machine peut posséder plusieurs capteurs
.
La table Incident intègre des clés étrangères vers Operateur et Machine pour refléter les liens hiérarchiques et d'association définis dans le schéma
.
La table Mesure est rattachée au Capteur correspondant
.
Types de données : Les types string, int, short, double et datetime ont été convertis vers leurs équivalents SQL standards (VARCHAR, INTEGER, SMALLINT, DOUBLE PRECISION, TIMESTAMP)

```python
from sqlalchemy import create_engine, Column, Integer, BigInteger, String, Float, DateTime, ForeignKey, SmallInteger, Text
from sqlalchemy.orm import declarative_base, relationship, sessionmaker

Base = declarative_base()

class Machine(Base):
    __tablename__ = 'machine'
    id = Column(BigInteger, primary_key=True) # ID (PK long not null) [1]
    nom = Column(String, nullable=False)      # Nom (string not null) [1]
    
    capteurs = relationship("Capteur", back_populates="machine")
    incidents = relationship("Incident", back_populates="machine") # Nouveau lien [1]

class Operateur(Base):
    __tablename__ = 'operateur'
    id = Column(BigInteger, primary_key=True)
    nom = Column(String, nullable=False)
    incidents = relationship("Incident", back_populates="operateur")

class Capteur(Base):
    __tablename__ = 'capteur'
    id = Column(BigInteger, primary_key=True)
    nom = Column(String, nullable=False)
    type = Column(Integer)
    unit = Column(String)
    machine_id = Column(BigInteger, ForeignKey('machine.id'), nullable=False)
    
    machine = relationship("Machine", back_populates="capteurs")
    mesures = relationship("Mesure", back_populates="capteur")

class Incident(Base):
    __tablename__ = 'incident'
    id = Column(BigInteger, primary_key=True)
    date = Column(DateTime, nullable=False)    # Date (datetime not null) [1]
    severite = Column(SmallInteger)            # severite (short null) [1]
    commentaire = Column(Text)                 # Commentaire (string null) [1]
    quarter = Column(SmallInteger)             # quarter (short null) [1]
    
    # Relations d'après le nouveau schéma [1]
    machine_id = Column(BigInteger, ForeignKey('machine.id'), nullable=False)
    operateur_id = Column(BigInteger, ForeignKey('operateur.id'), nullable=False)
    
    machine = relationship("Machine", back_populates="incidents")
    operateur = relationship("Operateur", back_populates="incidents")

class Mesure(Base):
    __tablename__ = 'mesure'
    id = Column(BigInteger, primary_key=True)
    valeur = Column(Float)                     # Valeur (double null) [1]
    date = Column(DateTime)                    # Date (datetime null) [1]
    capteur_id = Column(BigInteger, ForeignKey('capteur.id'), nullable=False)
    
    capteur = relationship("Capteur", back_populates="mesures")

# Connexion à la base de données (exemple avec SQLite pour le test)
# engine = create_engine('sqlite:///indusense.db')

# --- CONNEXION POSTGRESQL DANS DOCKER ---

# Remplacez par vos identifiants Docker
USER = "myuser"
PASSWORD = "mypassword"
HOST = "localhost" # Ou le nom du service si le script est aussi dans Docker
PORT = "5432"
DB_NAME = "mydb"

DATABASE_URL = f"postgresql://{USER}:{PASSWORD}@{HOST}:{PORT}/{DB_NAME}"

engine = create_engine(DATABASE_URL)
# engine = create_engine("postgresql+psycopg://dbuser:dbpassword@localhost:5432/mydb")
# Création des tables dans PostgreSQL
Base.metadata.create_all(engine)
```
